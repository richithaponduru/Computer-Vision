```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
```

```python
#Function to display images
def show_comparison(Original,blur3,blur5,blur11):
  plt.figure(figsize=(12,10))

  #Original Image
  plt.subplot(2,2,1)
  plt.imshow(Original,cmap='gray')
  plt.title("Original Grayscale Image")
  plt.axis('off')

  #3x3 kernel
  plt.subplot(2,2,2)
  plt.imshow(blur3,cmap='gray')
  plt.title("Average Filter(3x3)")
  plt.axis('off')

  #5x5 Kernel
  plt.subplot(2,2,3)
  plt.imshow(blur5,cmap='gray')
  plt.title("Average Filter (5x5)")
  plt.axis('off')

  #11x11
  plt.subplot(2,2,4)
  plt.imshow(blur11,cmap='gray')
  plt.title("Average Filter(11x11)")
  plt.axis('off')

  plt.tight_layout()
  plt.show()
```

```python
#1.REAd THE IMAGE IN GRAYSCALE
#
img=cv2.imread('/content/1_-x3F59-3Ap0TaiJ7-fS4Sw.png',0)

if img is None:
  print("Error:Image not found")
else:
  #2.Define avg kernels of diff size(3X3,5X5,7X7)
  kernel_3X3=np.ones((3,3),np.float32)/9
  kernel_5X5=np.ones((5,5),np.float32)/25
  kernel_11X11=np.ones((11,11),np.float32)/121

  #3.Apply each kernel to image using 2D convultion operation
  #cv2.filter2D performs convultion operation
  blur3=cv2.filter2D(img,-1,kernel_3X3)
  blur5=cv2.filter2D(img,-1,kernel_5X5)
  blur11=cv2.filter2D(img,-1,kernel_11X11)

  #4.Display the original and blurred images for comparision
  show_comparison(img,blur3,blur5,blur11)
```
